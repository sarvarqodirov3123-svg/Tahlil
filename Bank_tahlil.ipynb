```python
corr = df[['age','duration','campaign','pdays','previous','emp.var.rate','cons.price.idx','cons.conf.idx','euribor3m','nr.employed']].corr()
sns.heatmap(corr, annot=True, cmap="YlGnBu",vmin=-1, vmax=1, center=0)
plt.show()
```

```python
plt.figure(figsize=(20,10))
x = df['job'].unique()
y = df.groupby('job')['job'].count().sort_values(ascending=False)
plt.title("Mijozlar kasb kesimida.",size=20)
plt.xlabel("Kasiblar", size=16)
plt.ylabel("Soni", size=16)
plt.xticks(rotation=90)
plt.grid(True)
bars = plt.bar(x, y, align='center', color='blue', edgecolor='black', alpha=1)
plt.bar_label(bars, labels=y , padding=3, fontsize=12, fontweight='bold')
plt.show()
```

```python
ortacha_vaqt = round(df["duration"].mean(),2).astype(str) + " sekund"
ortacha_vaqt
```

```python
plt.figure(figsize=(20,10))
x = df['month'].unique()
y = (df.groupby('month')['campaign'].sum().sort_values(ascending=False)/df["campaign"].sum())*100
plt.title("Oylar kisimida qo'ng'iroqlar ulushi foizlarda.",size=20)
plt.xlabel("Oylar", size=16)
plt.ylabel("Foiz", size=16)
plt.xticks(rotation=90)
plt.grid(True)
bars = plt.bar(x, y, align='center', color='blue', edgecolor='black', alpha=1)
plt.bar_label(bars, labels=round(y,2).astype(str) + "%" , padding=3, fontsize=12, fontweight='bold')
plt.show()
```
